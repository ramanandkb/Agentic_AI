# Streaming and Batch

#### most models can stream their output content while it is being generated. By displaying output progressively. streaming significantly inproves user experience, particularly for longer resposes. Calling stream() resturns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk inreal-time

##### response without stream

```python
model.invoke("How to get 24LPA pkg")
```

##### response with stream

```python
for chunk in model.stream("write 3 para on how to get 24LPA pkg "):
    print(chunk.text,end="",flush=True)
```

#### by using markdown library it will display as final output

```python
from IPython.display import display, Markdown

text = ""
output = display(Markdown(""), display_id=True)

for chunk in model.stream("how to get 24LPA pkg"):
    text += chunk.text
    output.update(Markdown(text))
```

## Batch

Batching a colection of independent requests to a model  can significantly improve performance and reduce sosts as the peocessing can be done in parallel

```python
responses=model.batch_as_completed([
    "what is the capital of India?",
    "what is the capital of USA?", 
    "what is the capital of UK?"],
    config={
        'max_concurrency':2,
    }
    )

for index, res in responses:
    print(index, res.content)
```
